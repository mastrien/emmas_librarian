# Emma's Librarian: abordagens `decided` e `planned` em vários modelos (Colab, GPU)

Roda os 72 casos de `core/harness/eval/cases.jsonl` em cada modelo, nas abordagens novas. Mesmo código, prompt, seed 42
e nota (v2) das execuções anteriores.

- **`decided`**: um modelo de decisão (Liquid AI d1-2B, 1,1 GB) lê o pedido e escolhe os 2 assuntos mais prováveis em
  menos de 1 s; o modelo de linguagem recebe só as ferramentas desses assuntos. Se o roteador falha, ou a tentativa
  restrita não responde ou chama ferramenta fora dos assuntos **e nada alterou a biblioteca**, o completo assume.
  No LFM3 local: 63/72, contra 61 do completo e 59 do roteado, com 28% menos tempo.
- **`planned`**: o modelo entrega um plano chamando `submit_plan`, executa uma tarefa por vez e responde. No LFM3 local o
  plano custa de 1150 a 2048 tokens de raciocínio e falhou em parte da fumaça; a pergunta é se outros modelos planejam melhor.

**O que mede e o que não mede.** Mede *eficácia*. Os tempos daqui **não valem** para a RX 580.

**Atenção aos casos.** `rot-1` e `mon-3` foram corrigidos em 2026-11-17 (veja `docs/avaliacao-modelos.md`): nesses dois
casos os resultados não são comparáveis com o Colab anterior (`colab-eficacia-*`).

Como usar:
1. Gere o zip do código (commit atual) e coloque na pasta `emmas-eval` do Drive, com o nome de `ZIP_NAME`:
   `git archive --format=zip -o emmas-harness-eval-v3.zip HEAD:emmas_librarian`
2. Ambiente de execução: GPU (L4 ou T4). Rode as células em ordem.
3. A célula de fumaça roda 4 casos no primeiro modelo, em cada abordagem: confira antes de seguir para a bateria.
4. A execução é **retomável**: se a sessão cair, rode a célula de configuração e as seguintes; o que já terminou é pulado.
5. Se o kernel reiniciar, rode a célula de configuração de novo (ela define tudo o que as outras usam).

```
# 2. CONFIGURAÇÃO (o que quiser mudar fica aqui; rode de novo se o kernel reiniciar)
import os, json, subprocess, glob, time, datetime, hashlib
ZIP_NAME = 'emmas-harness-eval-v3.zip'
DRIVE_DIR = '/content/drive/MyDrive/emmas-eval'      # o zip fica aqui; os resultados vão para DRIVE_DIR/runs
APP = '/content/emmas'
# Os candidatos que passaram de 70% no Colab anterior, mais dois que ficaram abaixo, para ver se o plano os ajuda.
MODELS = ['lfm3-1.7b', 'qwen3.5-4b', 'qwen3.5-9b', 'gemma-4-e4b', 'qwen3.5-2b', 'lfm2.5-8b-a1b']
MODES = ['decided', 'planned']   # decided primeiro: é o que mais promete. planned é lento (o plano gasta muitos tokens)
CASES = 'all'                    # 'all' ou ids separados por vírgula, ex.: 'art-2,proj-1'
SMOKE_CASES = 'proj-1,art-1,proj-5,ia-1'   # os 4 casos em que o LFM3 falhou/passou no teste local
LABEL = 'colab-abordagens'
LLAMA_REF = 'b12425'             # tag do llama.cpp. Se não compilar com a CUDA do Colab, troque por 'master': o commit real é gravado.
EXPECTED = None if CASES != 'all' else 72

# Modelos: id -> (repositório no Hugging Face, arquivo Q4_K_M). Os ids são os de modelCatalog.ts.
CATALOG = {
    'lfm3-1.7b': ('LiquidAI/LFM3-1.7B-GGUF', 'LFM3-1.7B-Q4_K_M.gguf'),
    'lfm2.5-8b-a1b': ('LiquidAI/LFM2.5-8B-A1B-GGUF', 'LFM2.5-8B-A1B-Q4_K_M.gguf'),
    'qwen3.5-2b': ('unsloth/Qwen3.5-2B-GGUF', 'Qwen3.5-2B-Q4_K_M.gguf'),
    'qwen3.5-4b': ('unsloth/Qwen3.5-4B-GGUF', 'Qwen3.5-4B-Q4_K_M.gguf'),
    'qwen3.5-9b': ('unsloth/Qwen3.5-9B-GGUF', 'Qwen3.5-9B-Q4_K_M.gguf'),
    'gemma-4-e4b': ('unsloth/gemma-4-E4B-it-GGUF', 'gemma-4-E4B-it-Q4_K_M.gguf'),
    'lfm2.5-1.2b-instruct': ('LiquidAI/LFM2.5-1.2B-Instruct-GGUF', 'LFM2.5-1.2B-Instruct-Q4_K_M.gguf'),
    'lfm2-1.2b-tool': ('LiquidAI/LFM2-1.2B-Tool-GGUF', 'LFM2-1.2B-Tool-Q4_K_M.gguf'),
    'lfm2.5-350m': ('LiquidAI/LFM2.5-350M-GGUF', 'LFM2.5-350M-Q4_K_M.gguf'),
}
PASS_THRESHOLD = 0.70
# Modelo de decisão do modo decided (o mesmo llama-server serve os dois, pelo endpoint /v1/systemone).
DECISION_MODEL = 'd1-2b'
DECISION_CATALOG = {'d1-2b': ('LiquidAI/d1-2B-GGUF', 'd1-2B-Q4_K_M.gguf')}
TOP_AREAS = 2                    # quantos assuntos o modelo de linguagem recebe
print('modelos:', MODELS, '| modos:', MODES, '| casos:', CASES)
```

```
# 6. llama.cpp compilado aqui, na versão LLAMA_REF (o binário pronto não funcionou com a CUDA 13 do Colab)
!rm -rf /content/llama
!git clone --depth 1 --branch {LLAMA_REF} https://github.com/ggml-org/llama.cpp.git /content/llama
print('Compilando llama.cpp com CUDA (alguns minutos)...')
!cd /content/llama && cmake -B build -DGGML_CUDA=ON 2>&1 | tail -3 && cmake --build build --config Release -t llama-server -j $(nproc) 2>&1 | tail -3
SERVER = '/content/llama/build/bin/llama-server'
os.environ['LD_LIBRARY_PATH'] = '/content/llama/build/bin:/usr/lib64-nvidia:' + os.environ.get('LD_LIBRARY_PATH', '')
COMMIT = subprocess.run(['git', '-C', '/content/llama', 'rev-parse', '--short', 'HEAD'], capture_output=True, text=True).stdout.strip()
LLAMA_BUILD = f'{LLAMA_REF}@{COMMIT}'          # o que vai para o registro de cada execução
listed = subprocess.run([SERVER, '--list-devices'], capture_output=True, text=True)
print(listed.stdout + listed.stderr)
import re
found = re.search(r'^\s*(CUDA\d+):', listed.stdout + listed.stderr, re.M)
assert found, 'llama.cpp NÃO enxerga a GPU e rodaria na CPU: confira a célula 1 e a saída acima.'
help_text = subprocess.run([SERVER, '--help'], capture_output=True, text=True)
assert 'decision' in (help_text.stdout + help_text.stderr).lower(), f'Este llama.cpp ({LLAMA_REF}) não parece servir modelos de decisão: use um LLAMA_REF mais novo.'
DEVICE = found.group(1)
print('build', LLAMA_BUILD, '| dispositivo', DEVICE)
```

```
# 7. Modelos (direto do Hugging Face para o disco do Colab; o hash e o tamanho vão para o registro no Drive)
os.makedirs('/content/models', exist_ok=True)

def sha256(path):
    h = hashlib.sha256()
    with open(path, 'rb') as f:
        for block in iter(lambda: f.read(1 << 24), b''):
            h.update(block)
    return h.hexdigest()

wanted = [CATALOG[m] for m in MODELS] + ([DECISION_CATALOG[DECISION_MODEL]] if 'decided' in MODES else [])
for repo, name in wanted:
    dest = f'/content/models/{name}'
    if os.path.exists(dest):
        continue
    started = time.time()
    subprocess.run(['curl', '-L', '--fail', '-sS', '--retry', '3', '-o', dest + '.part', f'https://huggingface.co/{repo}/resolve/main/{name}'], check=True)
    os.rename(dest + '.part', dest)       # só existe com o nome final se o download terminou
    record = {'at': datetime.datetime.now().isoformat(timespec='seconds'), 'what': f'{repo}/{name}', 'bytes': os.path.getsize(dest),
              'seconds': round(time.time() - started, 1), 'sha256': sha256(dest)}
    open(f'{DRIVE_DIR}/runs/setup-colab.jsonl', 'a').write(json.dumps(record) + '\n')
    print(record['what'], record['bytes'] // 10**6, 'MB', record['seconds'], 's')
```

```
# 8. A função que roda um modelo num modo (a mesma na fumaça e na bateria)
INDEX = f'{DRIVE_DIR}/runs/index.jsonl'

def already_done(model, mode):
    if not os.path.exists(INDEX):
        return False
    for line in open(INDEX):
        d = json.loads(line)
        if d.get('model') == model and d.get('mode') == mode and d.get('label') == f'{LABEL}-{mode}':
            return EXPECTED is None or d.get('cases') == EXPECTED
    return False

ENV = dict(os.environ, LLAMA_SERVER=SERVER, LLAMA_MODELS_DIR='/content/models', LLAMA_DEVICE=DEVICE,
           LLAMA_BACKEND='cuda', LLAMA_BUILD=LLAMA_BUILD, EVAL_RUNS_DIR=f'{DRIVE_DIR}/runs')

def run_eval(model, mode, cases, label):
    extra = ['--decision-model', DECISION_MODEL, '--top-areas', str(TOP_AREAS)] if mode == 'decided' else []
    process = subprocess.Popen(['npx', 'vite-node', 'core/harness/eval/run/runEval.ts', '--', '--model', model, '--cases', cases,
                                '--mode', mode, '--label', label, *extra], cwd=APP, env=ENV,
                               stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    log = open(f'{DRIVE_DIR}/runs/log-{label}-{model}.txt', 'w')
    for line in process.stdout:
        log.write(line)
        if line.startswith(('PASS', 'FAIL')) or 'Error' in line:
            print(line.rstrip())
    process.wait()
    log.close()
    print('saída', process.returncode)
```

```
# 9. Fumaça: 4 casos no primeiro modelo, em cada abordagem. O plano sai? O roteador acerta? (rode antes da bateria)
for mode in MODES:
    run_eval(MODELS[0], mode, SMOKE_CASES, f'fumaca-{mode}')
    last = sorted(glob.glob(f'{DRIVE_DIR}/runs/2026-*_{MODELS[0]}'))[-1]
    print('=====', mode, last)
    for path in sorted(glob.glob(f'{last}/episodes/*.json')):
        e = json.load(open(path, encoding='utf-8'))
        r = e['record']
        info = r.get('decision') or {'plano': r.get('plan')}
        print(e['id'], 'passou' if e['score']['pass'] else 'falhou', '| parou:', r['stopped'], '| ferramentas:', e['toolCount'], '|', info, '|', (r.get('error') or '')[:100])
```

```
# 10. A bateria: cada modelo, cada modo. Retomável: o que já está no índice do Drive é pulado.
for mode in MODES:
    for model in MODELS:
        if already_done(model, mode):
            print('já feito:', model, mode)
            continue
        print(f'=== {model} / {mode}')
        run_eval(model, mode, CASES, f'{LABEL}-{mode}')
```

```
# 11. Resultado: as abordagens novas contra as anteriores (roteado e completo, do Colab anterior, se estiver no mesmo índice)
import pandas as pd
rows = []
for line in open(INDEX):
    d = json.loads(line)
    label = str(d.get('label', ''))
    if d.get('cases') != 72 or not (label.startswith(LABEL) or label.startswith('colab-eficacia')):
        continue
    rows.append({'modelo': d['model'], 'abordagem': d['mode'], 'passou': d['passed'], 'taxa': round(d['passed'] / d['cases'], 2),
                 'minutos (não vale p/ a RX 580)': round(d['totalWallMs'] / 60000, 1)})
table = pd.DataFrame(rows).drop_duplicates(['modelo', 'abordagem'], keep='last')
display(table.pivot(index='modelo', columns='abordagem', values='taxa'))
for mode in MODES:
    done = table[table['abordagem'] == mode].set_index('modelo')['taxa']
    print('Com %s, passam de %.0f%%:' % (mode, PASS_THRESHOLD * 100), sorted(done[done >= PASS_THRESHOLD].index))
print('Lembrete: rot-1 e mon-3 mudaram; nesses dois casos o Colab anterior não é comparável.')
```

```
# 12. Em detalhe: o plano (planned) e o roteador (decided)
plan_stats, decided_stats = [], []
for config_path in sorted(glob.glob(f'{DRIVE_DIR}/runs/2026-*/config.json')):
    config = json.load(open(config_path, encoding='utf-8'))
    if not str(config.get('label') or '').startswith(LABEL):
        continue
    folder = os.path.dirname(config_path)
    episodes = [json.load(open(p, encoding='utf-8')) for p in glob.glob(f'{folder}/episodes/*.json')]
    model = config['model']['id']
    if config.get('mode') == 'planned':
        tokens = [e['record']['steps'][0]['reply']['timings']['generatedTokens'] for e in episodes if e['record']['steps']]
        capped = sum(e['record']['steps'][0]['reply']['finishReason'] == 'length' for e in episodes if e['record']['steps'])
        invalid = sum('ERR_INVALID_PLAN' in (e['record'].get('error') or '') for e in episodes)
        empty = sum(len(e['record'].get('plan') or [1]) == 0 for e in episodes)
        plan_stats.append({'modelo': model, 'casos': len(episodes), 'tokens do plano (média)': round(sum(tokens) / max(len(tokens), 1)),
                           'bateu no limite': capped, 'plano inválido': invalid, 'plano vazio': empty})
    if config.get('mode') == 'decided':
        traces = [e['record']['decision'] for e in episodes if e['record'].get('decision')]
        fallbacks = {}
        for t in traces:
            fallbacks[t['fallback']] = fallbacks.get(t['fallback'], 0) + 1
        decided_stats.append({'modelo': model, 'casos': len(episodes), 'reservas no completo': fallbacks,
                              'ferramentas oferecidas (média)': round(sum(e['toolCount'] for e in episodes) / max(len(episodes), 1), 1),
                              'decisão (ms, média)': round(sum(t['decisionMs'] for t in traces) / max(len(traces), 1))})
print('planned:'); display(pd.DataFrame(plan_stats))
print('decided:'); display(pd.DataFrame(decided_stats))
```
